```python
import transformers
from transformers.pipelines import SUPPORTED_TASKS

print(transformers.__version__)
print("summarization" in SUPPORTED_TASKS)
```

```python
import os
os.environ["USE_TF"] = "0"

from transformers import pipeline

summarizer = pipeline(
    "summarization",
    model="sshleifer/distilbart-cnn-12-6",
    framework="pt"
)

print("BART summarization model loaded successfully!")
```

```python
summary = summarizer(
    transcript,
    max_length=150,
    min_length=40,
    do_sample=False,
    truncation=True
)

meeting_summary = summary[0]["summary_text"]

print("Meeting Summary:")
print(meeting_summary)
```

```python
with open("transcript.txt", "w", encoding="utf-8") as f:
    f.write(transcript)

with open("meeting_summary.txt", "w", encoding="utf-8") as f:
    f.write(meeting_summary)

print("Transcript and summary saved successfully!")
```

```python
from transformers import pipeline

keypoint_extractor = pipeline(
    "summarization",
    model="sshleifer/distilbart-cnn-12-6",
    framework="pt"
)

key_points = keypoint_extractor(
    transcript,
    max_length=100,
    min_length=25,
    do_sample=False,
    truncation=True
)

print("KEY DISCUSSION POINTS:")
print(key_points[0]["summary_text"])
```

```python
import re

sentences = re.split(r'(?<=[.!?])\s+', transcript)

keywords = [
    "need to", "should", "will", "must",
    "action", "follow up", "next step",
    "responsible for", "deadline", "assign"
]

action_items = [
    sentence.strip()
    for sentence in sentences
    if any(word in sentence.lower() for word in keywords)
]

print("POSSIBLE ACTION ITEMS:")
for item in action_items:
    print("-", item)
```

```python
with open("meeting_report.txt", "w", encoding="utf-8") as f:
    f.write("MEETING SUMMARY\n")
    f.write(meeting_summary + "\n\n")

    f.write("KEY DISCUSSION POINTS\n")
    f.write(key_points[0]["summary_text"] + "\n\n")

    f.write("POSSIBLE ACTION ITEMS\n")
    for item in action_items:
        f.write("- " + item + "\n")

print("Meeting report saved successfully!")
```
